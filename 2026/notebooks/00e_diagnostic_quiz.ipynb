```python
import collections

by_topic = collections.defaultdict(lambda: [0, 0])
for item, ok in RESULTS.items():
    by_topic[TOPICS[item]][1] += 1
    by_topic[TOPICS[item]][0] += int(ok)

total_ok = sum(RESULTS.values())
total = len(RESULTS)
pct = 100 * total_ok / total if total else 0

print("=" * 58)
print(f"DIAGNOSTIC REPORT - {total_ok}/{total} correct ({pct:.0f}%)")
print("=" * 58)
for topic, (ok, n) in sorted(by_topic.items(), key=lambda kv: kv[1][0] / kv[1][1]):
    bar = "#" * ok + "." * (n - ok)
    flag = "   <-- work on this" if ok < n else ""
    print(f"  {topic:<20s} {ok}/{n}  {bar}{flag}")

print()
if pct >= 70:
    print("You are ready for Session 1.")
    weak = [t for t, (ok, n) in by_topic.items() if ok < n]
    if weak:
        print(f"Still worth reviewing before we start: {', '.join(weak)}.")
else:
    print("Below the 70% threshold. Work through 00f_Remediation, focusing on the")
    print("topics flagged above, then retake this quiz. This is normal and fixable -")
    print("tell me now rather than in week 5.")
print()
print("Send me this report (the whole block), not just the percentage.")
```
