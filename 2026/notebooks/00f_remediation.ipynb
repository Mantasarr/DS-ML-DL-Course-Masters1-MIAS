## §6 - Comprehensions ⭐ *priority section*

### The idea

A comprehension builds a list in one expression. There are **two different things**
it can do, and conflating them is the single most expensive mistake in this notebook.

```python
[v * 2      for v in values]                # MAP:       same length, values changed
[v          for v in values if v > 0]       # FILTER:    shorter, values unchanged
[v if v > 0 else 0  for v in values]        # TRANSFORM: same length, values changed
```

| where the condition goes | what it does | length |
|---|---|---|
| `if` at the **end** | filters - drops elements | **changes** |
| `a if cond else b` at the **front** | transforms - replaces elements | **preserved** |

Why this matters more than it looks: in pandas, using a filter where you meant a
transform silently changes your row count. Every number computed afterwards is then
wrong, and **nothing raises an error**. That is far more expensive than a crash.

Dict comprehensions use the same shape with `{key: value for ...}`.

## §8 - Reading errors ⭐ *priority section*

### The idea

**Read a traceback from the bottom.**

1. The **last line** says *what* went wrong.
2. The **last `File ... line N`** block above it says *where* - that is the line to
   look at.
3. Everything above that is *how you got there*: the chain of calls. Useful context,
   not the bug.

Almost everyone reads the top first and goes to look at the wrong line.

### The five you will actually meet

| error | what it means |
|---|---|
| `KeyError: 'x'` | no such dict key or DataFrame column - usually a typo |
| `TypeError: unsupported operand ...` | wrong types, e.g. a string where a number belongs |
| `ValueError: could not convert ...` | right type, impossible value - `float('$409')` |
| `AttributeError: 'NoneType' has no attribute ...` | you used the result of something that returned `None` |
| `IndexError: list index out of range` | index past the end, or an empty list |
